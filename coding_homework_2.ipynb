```python
# import libraries

import numpy as np
import matplotlib.pyplot as plt
import statistics as st
```

```python
# import csvs & create arrays

v_rest1 = np.loadtxt('https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/v_rest_sst.csv')
v_rest2 = np.loadtxt('https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/v_rest_pvalb.csv')
```

```python
# concatenate datasets into smaller matrix

# stacks 13 columns from each dataset on top of each other
v_rest = np.vstack((v_rest1[:13], v_rest2[:13]))

print(v_rest.shape)
```

```python
# print size of v-rest

np.size(v_rest)
```

```python
# return first row
v_rest[0]
```

```python
# return sum of vals in 3rd column
print(v_rest[:, 2].sum())
```

```python
# return second to last value in 2nd row v_rest
print(v_rest[1][-2])
```

```python
# make histogram of vals in 1st row v_rest

plt.hist(v_rest[0])
plt.xlabel('Resting membrane potential')
plt.ylabel('Frequency')
```

```python
# find mean RMP in sst and pvalb, assign means to new vars

v_rest_mean_sst = st.mean(v_rest[0])
v_rest_mean_pvalb = st.mean(v_rest[1])

print(v_rest_mean_sst)
print(v_rest_mean_pvalb)
```

##### Sst has a lower/more negative resting membrane potential than pvalb in this sample.

```python
# import scipy pkg

import scipy.stats as stats
```

```python
# do t-test comparing difference between mean sst and pvalb RMPs

stats.ttest_ind(v_rest[0], v_rest[1])
```

My p-value was *p* = 0.65. Since *p* > 0.05, the means for sst and pvalb were not significantly different. However, I used a convenience sample from each dataset (the first 13 datapoints), which is most likely not representative of the full range of RMPs for sst and pvalb. Using random sampling would be preferable to get less biased measurements of RMPs to more confidently generalize these results to the whole dataset.
