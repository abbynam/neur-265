```python
# import libraries

import numpy as np
import matplotlib.pyplot as plt
import statistics as st
```

```python
# import csvs & create arrays

v_rest1 = np.loadtxt('https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/v_rest_sst.csv')
v_rest2 = np.loadtxt('https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/v_rest_pvalb.csv')
```

```python
# concatenate datasets into matrix
v_rest = np.array([v_rest1, v_rest2])
```

```python
# check shape & print size of v-rest

print(v_rest.shape)
print(np.size(v_rest))
```

```python
# return first row (sst)
v_rest[0]
```

```python
# return sum of vals in 3rd column
print(v_rest[:, 2].sum())
```

```python
# return second to last value in 2nd row v_rest
print(v_rest[1, -2])
```

```python
# make histogram of vals in 1st row v_rest

plt.hist(v_rest[0])
plt.xlabel('Resting membrane potential (mV)')
plt.ylabel('Frequency')
```

```python
# find mean RMP in sst and pvalb, assign means to new vars

v_rest_mean_sst = st.mean(v_rest[0])
v_rest_mean_pvalb = st.mean(v_rest[1])

print(v_rest_mean_sst)
print(v_rest_mean_pvalb)
```

##### Pvalb has a lower/more negative mean resting membrane potential than sst in this sample.

```python
# import scipy pkg

import scipy.stats as stats
```

```python
# do t-test comparing difference between mean sst and pvalb RMPs

stats.ttest_ind(v_rest[0], v_rest[1])
```

My p-value was *p* = 7.159828660199036e-09, which is a very small number. Since *p* < 0.05, the mean RMPs for sst and pvalb neurons were significantly different.
