```python
# import libraries

import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats
```

```python
# import v_rest csvs as numpy arrays

v_rest_sst = np.loadtxt('https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/v_rest_sst.csv')
v_rest_pvalb = np.loadtxt('https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/v_rest_pvalb.csv')
```

```python
# find means
print(np.mean(v_rest_sst))
print(np.mean(v_rest_pvalb))

# run a t-test
t_result = stats.ttest_ind(v_rest_sst, v_rest_pvalb)

# plot histogram
if t_result.pvalue < 0.05:
    fig, ax = plt.subplots(1, 2)

    ax[0].hist(v_rest_sst)
    ax[1].hist(v_rest_pvalb)

    ax[0].set_xlabel('Resting membrane potential (mV)')
    ax[1].set_xlabel('Resting membrane potential (mV)')

    ax[0].set_ylabel('Frequency')
    ax[1].set_ylabel('Frequency')

    plt.tight_layout()
```

```python
# define function
def vrestHist():
  """
  This function creates 2 histograms to show the
  frequency of Sst and Pvalb neurons' resting membrane
  potentials.
  """

  if t_result.pvalue < 0.05:
    fig, ax = plt.subplots(1, 2)

    ax[0].hist(v_rest_sst)
    ax[1].hist(v_rest_pvalb)

    ax[0].set_xlabel('Resting membrane potential (mV)')
    ax[1].set_xlabel('Resting membrane potential (mV)')

    ax[0].set_ylabel('Frequency')
    ax[1].set_ylabel('Frequency')

    plt.tight_layout()
```

```python
# call function
vrestHist()
```

```python
#  Loop thru e/ column of v_rest sst and pvalb
v_rest_length = len(v_rest_sst)

for i in range(v_rest_length):
  if v_rest_sst[i] > -60 and v_rest_pvalb[i] > -60:
    plt.scatter(x=v_rest_sst[i], y=v_rest_pvalb[i], color="red")
  elif v_rest_sst[i] > -60 and v_rest_pvalb[i] < -60:
    plt.scatter(x=v_rest_sst[i], y=v_rest_pvalb[i], color="green")
  elif v_rest_sst[i] < -60 and v_rest_pvalb[i] < -60:
    plt.scatter(x=v_rest_sst[i], y=v_rest_pvalb[i], color="blue")
  elif v_rest_sst[i] < -60 and v_rest_pvalb[i] > -60:
    plt.scatter(x=v_rest_sst[i], y=v_rest_pvalb[i], color="purple")

plt.xlabel('RMP of Sst (mV)')
plt.ylabel('RMP of Pvalb (mV)')
```

```python
# define function for scatterplot
def vrestScatter():
  """
  Creates a scatterplot with the Sst-expressing neurons' RMP on the x-axis and
  Pvalb-expressing neurons' RMP on the y-axis, coloring each datapoint based
  on how the RMPs compare to each other.
  """
  for i in range(v_rest_length):
    if v_rest_sst[i] > -60 and v_rest_pvalb[i] > -60:
      plt.scatter(x=v_rest_sst[i], y=v_rest_pvalb[i], color="red")
    elif v_rest_sst[i] > -60 and v_rest_pvalb[i] < -60:
      plt.scatter(x=v_rest_sst[i], y=v_rest_pvalb[i], color="green")
    elif v_rest_sst[i] < -60 and v_rest_pvalb[i] < -60:
      plt.scatter(x=v_rest_sst[i], y=v_rest_pvalb[i], color="blue")
    elif v_rest_sst[i] < -60 and v_rest_pvalb[i] > -60:
      plt.scatter(x=v_rest_sst[i], y=v_rest_pvalb[i], color="purple")

  plt.xlabel('RMP of Sst (mV)')
  plt.ylabel('RMP of Pvalb (mV)')
```

```python
# call function for scatterplot
vrestScatter()
```
