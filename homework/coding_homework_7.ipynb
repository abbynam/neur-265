```python
# import modules

import numpy as np
import matplotlib.pyplot as plt
```

```python
# load csvs as numpy arrays

# false alarms (hit incorrect stimulus)
acc_fa = np.loadtxt('https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/acc_fa_hw5.csv', delimiter=',')
# hits (hit correct stimulus)
acc_hits = np.loadtxt('https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/acc_test.csv', delimiter = ',')
```

```python
# print shape of each array

print(np.shape(acc_fa))
print(np.shape(acc_hits))
```

```python
# create time variable
time = np.linspace(0, 4, 8001)
```

```python
# calculate sampling rate
sampling_rate = 8001/4
print(sampling_rate)
```

```python
# subplots for hits and false alarms
fig, ax = plt.subplots(1,2)

# hits
ax[0].plot(time, acc_hits[0, :])
ax[0].set_xlabel('time [s]')
ax[0].set_ylabel('voltage [$\mu V$]')

# false alarms
ax[1].plot(time, acc_fa[0, :])
ax[1].set_xlabel('time [s]')
ax[1].set_ylabel('voltage [$\mu V$]')

plt.tight_layout()
```

It's very hard to tell what the dominant oscillation is of each graph without using a power spectrum for both. From just these graphs alone, the dominant oscillation seems slightly different between the first trials of acc_hits and acc_fa. I think the dominant oscillation of acc_hits trial 1 is around 1-5 Hz, and the dominant oscillation of acc_fa trial 1 is around 1-2 Hz. This trial of acc_hits is very noisy, so it's hard to see how and where the wave is cycling exactly.

```python
# calculate and plot ERP for acc_hits

plt.plot(time, acc_hits.mean(0))
plt.xlabel('time [s]')
plt.ylabel('voltage [$\mu V$]')
plt.vlines(2, -35, 35, color = 'black')
plt.hlines(0, time[0], time[-1], color = 'black')
```

The activity in the LFP is somewhat consistent throughout this trial, since the ERP seems to deviate a lot from the horizontal line. There is a change in the wave form right before the screen touch with a higher amplitude (more negative) trough, and right after the screen touch with a broader wave and higher amplitude (more positive) peak. In other words, the biggest difference in voltage happens directly before and after the mouse touched the screen.

```python
# calculate and plot ERP for acc_fa

plt.plot(time, acc_fa.mean(0))
plt.xlabel('time [s]')
plt.ylabel('voltage [$\mu V$]')
plt.vlines(2, -10, 10, color = 'black')
plt.hlines(0, time[0], time[-1], color = 'black')
```

The ERP for the false alarms is definitely less noisy, as the peaks and troughs are clearer. The magnitude of changes in voltage is lower than for the hits, but the overall pattern observed just before and after the mouse touched the screen was similar — the difference between the trough and peak of the wave was much higher in amplitude than previous and subsequent cycles of the wave. Overall, the wave seems to start at a high voltage, then progressively gets lower with time, or until it is interrupted by the mouse responding to the stimulus.
