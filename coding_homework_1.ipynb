```python
# install pandas and numpy

import pandas as pd
import numpy as np
from matplotlib import pyplot as plt
```

```python
# turns v_rest_sst.csv into pandas dataframe called pandas_data1

url = 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/v_rest_sst.csv'
pandas_data1 = pd.read_csv(url)
```

```python
# turns v_rest_pvalb.csv into pandas dataframe

url = 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/v_rest_pvalb.csv'
pandas_data2 = pd.read_csv(url)
```

```python
# converts sst and pvalb datasets into list variables

data1 = pandas_data1.values.tolist()
data2 = pandas_data2.values.tolist()
```

```python
# uses squeeze function so each element isn't bracketed

data1 = np.squeeze(data1)
data2 = np.squeeze(data2)
```

```python
# prints length of data1

print(len(data1))
```

```python
# print 3rd, 4th, 5th, 6th elements from data2

print(data2[2], data2[3], data2[4], data2[5])
```

```python
# print 2nd to last and last elements of data1

print(data1[-1], data1[-2])
```

```python
# print type of data2

type(data2)
```

```python
# make a histogram of data1 using matplotlib

plt.hist(data1)
```

```python
# remake histogram and label axes for data1


plt.hist(data1)
plt.xlabel('Resting membrane potential')
plt.ylabel('Frequency')


# changed limits of x and y axis to better compare between histograms
plt.xlim(-90, -40)
plt.ylim(0, 160)
```

```python
# create histogram for data2

plt.hist(data2)
```

```python
# remake histogram with axis labels for data2

plt.hist(data2)
plt.xlabel('Resting membrane potential')
plt.ylabel('Frequency')

# changed limits of x and y axis to better compare between histograms
plt.xlim(-90, -40)
plt.ylim(0, 160)
```
